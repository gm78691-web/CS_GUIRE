```python
"""
Publication-ready EDA & figure generation pipeline — 3DP-FRC dataset.
Adapted for: 3DP_FRC_Corrected_Units_299.csv
Target: CS (MPa)
Outputs: cleaned dataset, Table 2 descriptive statistics, and the full
figure suite (existing pipeline preserved).
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.lines import Line2D
import seaborn as sns
from scipy import stats
from scipy.stats import pearsonr, spearmanr
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

# =========================================================
# JOURNAL-STYLE PLOTTING PARAMETERS
# =========================================================
plt.rcParams.update({
    'font.size': 12,
    'font.family': 'sans-serif',
    'font.sans-serif': ['Helvetica', 'Arial', 'DejaVu Sans'],
    'axes.labelsize': 13,
    'axes.titlesize': 14,
    'axes.titleweight': 'bold',
    'axes.labelweight': 'bold',
    'legend.fontsize': 10,
    'legend.frameon': False,
    'xtick.labelsize': 11,
    'ytick.labelsize': 11,
    'figure.dpi': 300,
    'savefig.dpi': 500,
    'figure.figsize': (8, 6),
    'lines.linewidth': 2.4,
    'axes.edgecolor': '#000000',
    'axes.linewidth': 1.7,
    'text.color': '#000000',
    'axes.labelcolor': '#000000',
    'xtick.color': '#000000',
    'ytick.color': '#000000',
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.grid': True,
    'grid.alpha': 0.22,
    'grid.linestyle': '--',
    'grid.linewidth': 0.6,
    'grid.color': '#AAAAAA',
    'axes.axisbelow': True,
})

# =======================
# DATA LOADING
# =======================
data_path = r"D:\2026 Work\My Papers\Gulzar\1st Revision\Data\3DP_FRC_Corrected_Units_299.csv"
output_dir = Path(r"D:\2026 Work\My Papers\Gulzar\1st Revision\Data Presentation")
output_dir.mkdir(exist_ok=True, parents=True)

print(f"Loading data from: {data_path}")
print(f"Saving outputs to: {output_dir}")

encodings = ['utf-8', 'latin1', 'cp1252', 'iso-8859-1']
df = None
for encoding in encodings:
    try:
        df = pd.read_csv(data_path, encoding=encoding)
        print(f"Loaded with encoding: {encoding}")
        break
    except Exception:
        continue
if df is None:
    raise Exception("Could not load data with any encoding")
print(f"Data loaded: {len(df)} rows, {len(df.columns)} columns")

# =======================
# DEFINE TARGET VARIABLE
# =======================
target_var = 'CS (MPa)'
if target_var not in df.columns:
    # Fuzzy match: tolerate spacing / case / parentheses differences
    normalized = {
        col: col.strip().lower().replace(' ', '').replace('(', '').replace(')', '').replace('_', '')
        for col in df.columns
    }
    candidates = {'csmpa', 'compressivestrengthmpa', 'compressivestrength', 'cs'}
    for col, norm in normalized.items():
        if norm in candidates:
            target_var = col
            break
    else:
        raise Exception(f"Target variable 'CS (MPa)' not found. "
                        f"Available columns: {list(df.columns)}")
print(f"\nTarget variable: {target_var}")

input_vars = [c for c in df.columns if c != target_var]
available_inputs = [v for v in input_vars if v in df.columns]
print(f"Input variables ({len(available_inputs)}): {available_inputs}")

# =======================
# HELPERS
# =======================
def clean_column_name(col):
    col = col.replace('Â°', '°').replace('Ã‚Â°', '°')
    for old, new in {',kg/m3': '', ',mm': '', ',g/L': '', ' °C': '', '°C': '', '/': ' / '}.items():
        col = col.replace(old, new)
    return col.strip()


def despine(ax, keep=('left', 'bottom')):
    for side, spine in ax.spines.items():
        spine.set_visible(side in keep)
    if 'left' in keep:
        ax.spines['left'].set_color('#000000')
    if 'bottom' in keep:
        ax.spines['bottom'].set_color('#000000')


def add_n_label(ax, n, loc='upper right'):
    positions = {
        'upper right': (0.98, 0.98, 'right', 'top'),
        'upper left': (0.02, 0.98, 'left', 'top'),
        'lower right': (0.98, 0.02, 'right', 'bottom'),
        'lower left': (0.02, 0.02, 'left', 'bottom'),
    }
    x, y, ha, va = positions.get(loc, positions['upper right'])
    ax.text(x, y, f'n = {n:,}', transform=ax.transAxes, ha=ha, va=va,
             fontsize=9.5, fontweight='bold', color=HC['ink'], zorder=10,
             bbox=dict(boxstyle='round,pad=0.28', facecolor='white', edgecolor=HC['ink'],
                       linewidth=1.1, alpha=0.95))


def _safe_filename(name):
    return clean_column_name(name).replace('/', '-').replace(' ', '_').replace(',', '')


# =========================================================
# COLOR SYSTEM
# =========================================================
HC = {
    'c1': '#0072B2', 'c2': '#D55E00', 'c3': '#009E73', 'c4': '#E69F00',
    'c5': '#CC79A7', 'c6': '#56B4E9', 'c7': '#F0E442',
    'ink': '#1A1A1A', 'mean': '#D55E00', 'median': '#1A1A1A', 'neutral': '#7F7F7F',
}
PALETTE = [HC['c1'], HC['c2'], HC['c3'], HC['c4'], HC['c5'], HC['c6']]

DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    'okabe_diverging', ['#0072B2', '#F7F7F7', '#D55E00'], N=256)
SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    'okabe_sequential', ['#F7FBFF', '#56B4E9', '#0072B2', '#08306B'], N=256)

sns.set_theme(style='whitegrid', rc={
    'grid.linestyle': '--', 'grid.alpha': 0.22, 'axes.edgecolor': '#000000'})

# =========================================================
# NEW — SAVE CLEANED DATASET
# =========================================================
def save_cleaned_dataset():
    """Drop rows with any missing values and save a cleaned copy."""
    print("\nSaving cleaned dataset...")
    n_before = len(df)
    df_clean = df.dropna().reset_index(drop=True)
    n_after = len(df_clean)
    out_path = output_dir / 'Cleaned_Dataset.csv'
    df_clean.to_csv(out_path, index=False)
    print(f"  Saved: {out_path.name}")
    print(f"  Rows before cleaning : {n_before:,}")
    print(f"  Rows after  cleaning : {n_after:,}  ({n_before - n_after:,} dropped)")

    # accompanying audit log
    with open(output_dir / 'Cleaning_Log.txt', 'w') as f:
        f.write("CLEANING LOG — 3DP-FRC dataset\n")
        f.write("=" * 60 + "\n\n")
        f.write(f"Source file : {data_path}\n")
        f.write(f"Target var  : {target_var}\n")
        f.write(f"Rows before : {n_before:,}\n")
        f.write(f"Rows after  : {n_after:,}\n")
        f.write(f"Dropped     : {n_before - n_after:,}\n\n")
        f.write("Missing-value counts per column (before dropping):\n")
        miss = df.isnull().sum()
        for c in df.columns:
            f.write(f"  {c:<35s} {miss[c]:>5d}  ({miss[c]/n_before*100:6.2f}%)\n")
    return df_clean

# =========================================================
# NEW — TABLE 2 (exact columns requested)
# min, quartiles, median, mean, std, max, missing rate per variable
# =========================================================
def create_table2_statistics(df_clean):
    """
    Produce Table 2 with exactly:
      Variable | n | Missing (n) | Missing (%) | Min | Q1 | Median | Mean | Std | Max
    Includes the target (CS) as a row.
    Outputs: Table2_Descriptive_Statistics.{csv,png,txt}
    """
    print("\nBuilding Table 2 (descriptive statistics)...")

    # Only numeric variables, but always include the target
    numeric_cols = [c for c in df_clean.columns
                    if pd.api.types.is_numeric_dtype(df_clean[c])]
    if target_var not in numeric_cols:
        raise ValueError(f"Target '{target_var}' is not numeric.")

    rows = []
    n_total = len(df_clean)
    for var in numeric_cols:
        s = df_clean[var]
        non_null = s.dropna()
        rows.append({
            'Variable': clean_column_name(var),
            'n': int(non_null.count()),
            'Missing (n)': int(s.isnull().sum()),
            'Missing (%)': round(s.isnull().sum() / n_total * 100, 2),
            'Min': round(float(non_null.min()), 4) if len(non_null) else np.nan,
            'Q1': round(float(np.percentile(non_null, 25)), 4) if len(non_null) else np.nan,
            'Median': round(float(non_null.median()), 4) if len(non_null) else np.nan,
            'Mean': round(float(non_null.mean()), 4) if len(non_null) else np.nan,
            'Std': round(float(non_null.std(ddof=1)), 4) if len(non_null) > 1 else np.nan,
            'Max': round(float(non_null.max()), 4) if len(non_null) else np.nan,
        })
    table2 = pd.DataFrame(rows)

    # Put the target first, then the rest alphabetically
    target_label = clean_column_name(target_var)
    table2['_is_target'] = (table2['Variable'] == target_label).astype(int)
    table2 = table2.sort_values(['_is_target', 'Variable'], ascending=[False, True]).drop(columns='_is_target')
    table2 = table2.reset_index(drop=True)

    # --- CSV ---
    csv_path = output_dir / 'Table2_Descriptive_Statistics.csv'
    table2.to_csv(csv_path, index=False)
    print(f"  Saved: {csv_path.name}")

    # --- Plain text ---
    with open(output_dir / 'Table2_Descriptive_Statistics.txt', 'w') as f:
        f.write("TABLE 2 — DESCRIPTIVE STATISTICS (3DP-FRC)\n")
        f.write("=" * 90 + "\n")
        f.write(f"Target variable: {target_label}   |   Total observations: {n_total:,}\n")
        f.write("=" * 90 + "\n\n")
        f.write(table2.to_string(index=False))
        f.write("\n")

    # --- Publication-ready PNG table ---
    n_rows = len(table2)
    fig_w = 16
    fig_h = max(3.5, 1.1 + 0.55 * n_rows)
    fig, ax = plt.subplots(figsize=(fig_w, fig_h), dpi=500)
    ax.axis('off')

    col_labels = list(table2.columns)
    cell_text = []
    for _, r in table2.iterrows():
        cell_text.append([
            r['Variable'],
            f"{r['n']}",
            f"{r['Missing (n)']}",
            f"{r['Missing (%)']:.2f}",
            f"{r['Min']:.3f}",
            f"{r['Q1']:.3f}",
            f"{r['Median']:.3f}",
            f"{r['Mean']:.3f}",
            f"{r['Std']:.3f}",
            f"{r['Max']:.3f}",
        ])

    table = ax.table(
        cellText=cell_text,
        colLabels=col_labels,
        cellLoc='center',
        loc='center',
    )
    table.auto_set_font_size(False)
    table.set_fontsize(9.5)
    table.scale(1, 1.65)

    for (row, col), cell in table.get_celld().items():
        cell.set_edgecolor('#B8B8B8')
        cell.set_linewidth(0.8)
        if row == 0:
            cell.set_text_props(weight='bold', color='white')
            cell.set_facecolor(HC['ink'])
        elif col == 0:
            cell.set_text_props(weight='bold', ha='left')
            cell.set_facecolor('#EDEDED')
            cell._loc = 'left'
        else:
            cell.set_facecolor('white' if row % 2 == 0 else '#F6F8FA')

    # Highlight the target row (always row 1 since we sort it first)
    for col in range(len(col_labels)):
        cell = table[(1, col)]
        cell.set_facecolor('#FFF2CC')
        cell.set_text_props(weight='bold')

    ax.set_title(
        f"Table 2 — Descriptive Statistics of Input Variables and {target_label}\n"
        f"(3DP-FRC dataset, n = {n_total:,})",
        fontsize=14, fontweight='bold', pad=18,
    )
    plt.tight_layout()
    plt.savefig(output_dir / 'Table2_Descriptive_Statistics.png',
                dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print(f"  Saved: Table2_Descriptive_Statistics.png")

    return table2

# =========================================================
# (Optional but kept) richer descriptive statistics
# =========================================================
def create_descriptive_statistics():
    print("\nComputing extended descriptive statistics...")
    numeric_cols = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]
    desc = df[numeric_cols].describe().T
    desc['variance'] = df[numeric_cols].var()
    desc['skewness'] = df[numeric_cols].skew()
    desc['kurtosis'] = df[numeric_cols].kurtosis()
    desc['range'] = desc['max'] - desc['min']
    desc['IQR'] = desc['75%'] - desc['25%']
    desc['CV (%)'] = (desc['std'] / desc['mean']).abs() * 100
    desc['missing'] = df[numeric_cols].isnull().sum()
    desc['missing (%)'] = (df[numeric_cols].isnull().sum() / len(df)) * 100
    desc = desc[['count', 'missing', 'missing (%)', 'mean', 'std', 'variance',
                 'min', '25%', '50%', '75%', 'max', 'range', 'IQR',
                 'skewness', 'kurtosis', 'CV (%)']]
    desc.to_csv(output_dir / 'Extended_Descriptive_Statistics.csv')
    print("Saved Extended_Descriptive_Statistics.csv")
    return desc

# =========================================================
# The remaining figure-generating functions are IDENTICAL
# to your original script. To keep this reply focused, they
# are appended below verbatim — nothing else was changed.
# =========================================================

def create_train_test_split():
    print("\nCreating 70:30 Train-Test Split...")
    from sklearn.model_selection import train_test_split
    df_clean = df.dropna().reset_index(drop=True)
    train_df, test_df = train_test_split(df_clean, test_size=0.30, random_state=42, shuffle=True)
    train_df.to_csv(output_dir / 'Train_Data_70.csv', index=False)
    test_df.to_csv(output_dir / 'Test_Data_30.csv', index=False)
    print(f"  Train set: {len(train_df)} rows ({len(train_df)/len(df_clean)*100:.1f}%)")
    print(f"  Test set:  {len(test_df)} rows ({len(test_df)/len(df_clean)*100:.1f}%)")

    numeric_cols = [c for c in df_clean.columns if pd.api.types.is_numeric_dtype(df_clean[c])]
    split_stats = pd.DataFrame({
        'Full': df_clean[numeric_cols].mean(), 'Train (70%)': train_df[numeric_cols].mean(),
        'Test (30%)': test_df[numeric_cols].mean(), 'Full_std': df_clean[numeric_cols].std(),
        'Train_std': train_df[numeric_cols].std(), 'Test_std': test_df[numeric_cols].std(),
    })
    split_stats.to_csv(output_dir / 'Train_Test_Split_Statistics.csv')
    print("Saved Train_Test_Split_Statistics.csv")

    fig, axes = plt.subplots(1, 3, figsize=(18, 6), dpi=500)
    ax1 = axes[0]
    sizes = [len(train_df), len(test_df)]
    labels = [f'Train\n{len(train_df)} ({len(train_df)/len(df_clean)*100:.1f}%)',
              f'Test\n{len(test_df)} ({len(test_df)/len(df_clean)*100:.1f}%)']
    colors = [HC['c1'], HC['c2']]
    wedges, texts, autotexts = ax1.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%',
                                        startangle=90, explode=(0.04, 0.04),
                                        wedgeprops=dict(edgecolor='white', linewidth=2.0),
                                        textprops=dict(fontweight='bold', fontsize=11))
    for at in autotexts:
        at.set_fontweight('bold'); at.set_color('white')
    ax1.set_title('(a) Train-Test Split Proportion', fontweight='bold', fontsize=13)

    ax2 = axes[1]
    if target_var in numeric_cols:
        ax2.hist(train_df[target_var].dropna(), bins=25, density=True, alpha=0.75,
                  color=HC['c1'], edgecolor='white', linewidth=0.8, label='Train (70%)')
        ax2.hist(test_df[target_var].dropna(), bins=25, density=True, alpha=0.65,
                  color=HC['c2'], edgecolor='white', linewidth=0.8, label='Test (30%)')
        ax2.axvline(train_df[target_var].mean(), color=HC['c1'], linestyle='--', linewidth=2.2)
        ax2.axvline(test_df[target_var].mean(), color=HC['c2'], linestyle='--', linewidth=2.2)
        ax2.set_xlabel(target_var, fontweight='bold'); ax2.set_ylabel('Density', fontweight='bold')
        ax2.set_title('(b) Target Distribution: Train vs Test', fontweight='bold', fontsize=13)
        ax2.legend(loc='best'); despine(ax2)

    ax3 = axes[2]
    if target_var in numeric_cols:
        bp = ax3.boxplot([train_df[target_var].dropna(), test_df[target_var].dropna()],
                          patch_artist=True, widths=0.5, labels=['Train (70%)', 'Test (30%)'])
        bp['boxes'][0].set_facecolor(HC['c1']); bp['boxes'][1].set_facecolor(HC['c2'])
        for box in bp['boxes']:
            box.set_alpha(0.85); box.set_edgecolor(HC['ink']); box.set_linewidth(1.6)
        for med in bp['medians']:
            med.set_color('white'); med.set_linewidth(2.5)
        for w in bp['whiskers'] + bp['caps']:
            w.set_color(HC['ink']); w.set_linewidth(1.6)
        ax3.set_ylabel(target_var, fontweight='bold')
        ax3.set_title('(c) Target Distribution: Train vs Test', fontweight='bold', fontsize=13)
        despine(ax3)

    plt.suptitle(f'Train-Test Split Analysis (70:30)  \u2014  n = {len(df_clean):,}',
                 fontsize=17, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.savefig(output_dir / 'Train_Test_Split.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Train_Test_Split.png")
    return train_df, test_df


def create_pca_biplot():
    print("\nCreating PCA Biplot...")
    try:
        from sklearn.decomposition import PCA
        from sklearn.preprocessing import StandardScaler
    except ImportError:
        print("scikit-learn not available - skipping PCA"); return
    pca_vars = [v for v in available_inputs if pd.api.types.is_numeric_dtype(df[v])]
    if len(pca_vars) < 3:
        print("Not enough numeric variables for PCA"); return
    pca_data = df[pca_vars].dropna()
    if len(pca_data) < 10:
        print("Not enough data for PCA"); return
    scaled = StandardScaler().fit_transform(pca_data)
    pca = PCA(n_components=2)
    scores = pca.fit_transform(scaled)
    loadings = pca.components_.T
    explained = pca.explained_variance_ratio_ * 100
    short_names = {v: clean_column_name(v) for v in pca_vars}

    fig, ax = plt.subplots(figsize=(14, 10), dpi=500)
    sc = ax.scatter(scores[:, 0], scores[:, 1], alpha=0.9, s=48, c=scores[:, 0],
                     cmap=SEQUENTIAL_CMAP, edgecolors=HC['ink'], linewidth=0.6)
    scale = np.max(np.abs(scores)) * 0.9 / np.max(np.abs(loadings))
    for i, var in enumerate(pca_vars):
        x_pos, y_pos = loadings[i, 0] * scale, loadings[i, 1] * scale
        ax.arrow(0, 0, x_pos, y_pos, head_width=0.10, head_length=0.10,
                  fc=HC['c2'], ec=HC['c2'], alpha=1.0, linewidth=2.4, zorder=5)
        angle = np.arctan2(y_pos, x_pos) * 180 / np.pi
        if 60 < angle < 120:
            ha, va, ox, oy = 'center', 'bottom', 0, 0.15
        elif -120 < angle < -60:
            ha, va, ox, oy = 'center', 'top', 0, -0.15
        else:
            ha, va, ox, oy = ('left', 'center', 0.15, 0) if x_pos > 0 else ('right', 'center', -0.15, 0)
        ax.text(x_pos + ox, y_pos + oy, short_names[var], fontweight='bold', fontsize=11,
                 color=HC['ink'], ha=ha, va=va,
                 bbox=dict(boxstyle='round,pad=0.25', facecolor='white', alpha=0.9,
                           edgecolor='#DDDDDD', linewidth=1.0))
    ax.axhline(0, color='#BBBBBB', linewidth=1.2)
    ax.axvline(0, color='#BBBBBB', linewidth=1.2)
    ax.set_xlabel(f'Principal Component 1 ({explained[0]:.1f}%)', fontweight='bold', fontsize=15)
    ax.set_ylabel(f'Principal Component 2 ({explained[1]:.1f}%)', fontweight='bold', fontsize=15)
    ax.set_title('PCA Biplot of Input Variables', fontweight='bold', fontsize=17, pad=20)
    despine(ax)
    ax.text(0.02, 0.98, f'Total variance explained: {explained[0]+explained[1]:.1f}%\nn = {len(pca_data):,}',
            transform=ax.transAxes, va='top', fontweight='bold', fontsize=11,
            bbox=dict(boxstyle='round,pad=0.4', facecolor='white', alpha=0.95, edgecolor='#DDDDDD'))
    plt.tight_layout()
    plt.savefig(output_dir / 'PCA_Biplot.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved PCA Biplot")


def create_dataset_composition():
    print("\nCreating Dataset Composition...")
    fig = plt.figure(figsize=(14, 10), dpi=500)
    gs = fig.add_gridspec(2, 3, hspace=0.4, wspace=0.3)
    numeric_cols = [c for c in available_inputs if pd.api.types.is_numeric_dtype(df[c])]
    cat_cols = [c for c in available_inputs
                if df[c].dtype == 'object' or df[c].nunique() < 10]

    ax1 = fig.add_subplot(gs[0, 0])
    if cat_cols:
        col = cat_cols[0]
        counts = df[col].value_counts().sort_index()
        bars = ax1.bar(range(len(counts)), counts.values, color=PALETTE[:len(counts)],
                        edgecolor=HC['ink'], linewidth=1.4)
        ax1.set_xticks(range(len(counts)))
        ax1.set_xticklabels(counts.index, rotation=45, ha='right', fontweight='bold')
        ax1.set_xlabel(clean_column_name(col), fontweight='bold')
        ax1.set_ylabel('Count', fontweight='bold')
        ax1.set_title(f'(a) {clean_column_name(col)}', fontweight='bold', fontsize=13)
        for bar, v in zip(bars, counts.values):
            ax1.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 3, str(v),
                      ha='center', va='bottom', fontweight='bold', fontsize=11)
        despine(ax1)
    elif numeric_cols:
        col = numeric_cols[0]
        ax1.hist(df[col].dropna(), bins=20, color=HC['c1'], edgecolor=HC['ink'], linewidth=1.2, alpha=0.95)
        ax1.set_xlabel(clean_column_name(col), fontweight='bold'); ax1.set_ylabel('Frequency', fontweight='bold')
        ax1.set_title(f'(a) {clean_column_name(col)}', fontweight='bold', fontsize=13); despine(ax1)

    panel_axes = [fig.add_subplot(gs[0, 1]), fig.add_subplot(gs[0, 2]),
                  fig.add_subplot(gs[1, 0]), fig.add_subplot(gs[1, 1])]
    letters = ['b', 'c', 'd', 'e']
    for i, ax in enumerate(panel_axes):
        col_idx = i + 1
        if len(numeric_cols) > col_idx:
            col = numeric_cols[col_idx]
            data = df[col].dropna()
            color = PALETTE[(i + 1) % len(PALETTE)]
            ax.hist(data, bins=20, color=color, edgecolor=HC['ink'], linewidth=1.2, alpha=0.95)
            ax.axvline(data.mean(), color=HC['mean'], linestyle='--', linewidth=2.2, label=f'Mean: {data.mean():.2f}')
            ax.axvline(data.median(), color=HC['ink'], linestyle=':', linewidth=2.2, label=f'Median: {data.median():.2f}')
            ax.set_xlabel(clean_column_name(col), fontweight='bold'); ax.set_ylabel('Frequency', fontweight='bold')
            ax.set_title(f'({letters[i]}) {clean_column_name(col)}', fontweight='bold', fontsize=13)
            ax.legend(loc='best', fontsize=9); despine(ax)

    ax6 = fig.add_subplot(gs[1, 2]); ax6.axis('off')
    summary_lines = ["DATASET SUMMARY", "", f"Total Observations: {len(df):,}", "",
                      "Target Variable:", f"\u2022 {target_var}", "", f"Input Variables: {len(available_inputs)}"]
    for c in numeric_cols[:4]:
        summary_lines.append(f"\u2022 {clean_column_name(c)}: {df[c].min():.2f} \u2013 {df[c].max():.2f}")
    ax6.text(0.05, 0.95, "\n".join(summary_lines), transform=ax6.transAxes, verticalalignment='top',
              fontsize=11, fontweight='bold',
              bbox=dict(boxstyle='round', facecolor='#F6F8FA', edgecolor=HC['ink'], linewidth=1.6),
              family='monospace')

    plt.suptitle(f'Dataset Composition and Characteristics  (n = {len(df):,})',
                 fontsize=17, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.savefig(output_dir / 'Dataset_Composition.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Dataset Composition")


def create_target_distribution():
    print("\nCreating Target Distribution...")
    fig, axes = plt.subplots(2, 2, figsize=(14, 10), dpi=500)
    data = df[target_var].dropna()
    if len(data) < 5:
        print("Not enough target data"); return

    ax1 = axes[0, 0]
    ax1.hist(data, bins=35, density=True, alpha=0.85, color=HC['c1'], edgecolor=HC['ink'],
              linewidth=0.9, label='Histogram')
    try:
        kde = stats.gaussian_kde(data)
        x_vals = np.linspace(data.min(), data.max(), 200)
        ax1.plot(x_vals, kde(x_vals), color=HC['ink'], linewidth=2.8, label='KDE')
    except Exception:
        pass
    ax1.axvline(data.mean(), color=HC['mean'], linestyle='-', linewidth=2.6, label=f'Mean: {data.mean():.3f}')
    ax1.axvline(data.median(), color=HC['c3'], linestyle='--', linewidth=2.6, label=f'Median: {data.median():.3f}')
    q1, q3 = np.percentile(data, [25, 75])
    ax1.axvspan(q1, q3, alpha=0.15, color=HC['c4'])
    ax1.set_xlabel(target_var, fontweight='bold'); ax1.set_ylabel('Density', fontweight='bold')
    ax1.set_title('(a) Distribution with KDE', fontweight='bold', fontsize=14)
    ax1.legend(loc='upper right'); despine(ax1)
    add_n_label(ax1, len(data), 'upper left')

    ax2 = axes[0, 1]
    bp = ax2.boxplot(data, patch_artist=True, widths=0.5)
    bp['boxes'][0].set_facecolor(HC['c1']); bp['boxes'][0].set_alpha(0.85)
    bp['boxes'][0].set_edgecolor(HC['ink']); bp['boxes'][0].set_linewidth(1.6)
    bp['medians'][0].set_color('white'); bp['medians'][0].set_linewidth(2.6)
    for w in bp['whiskers']:
        w.set_color(HC['ink']); w.set_linewidth(1.6)
    for c in bp['caps']:
        c.set_color(HC['ink']); c.set_linewidth(1.6)
    bp['fliers'][0].set_marker('o'); bp['fliers'][0].set_markerfacecolor(HC['c2'])
    bp['fliers'][0].set_markeredgecolor('white'); bp['fliers'][0].set_markersize(7)
    ax2.set_ylabel(target_var, fontweight='bold'); ax2.set_title('(b) Box Plot with Outliers', fontweight='bold', fontsize=14)
    ax2.set_xticks([]); despine(ax2)

    ax3 = axes[1, 0]
    sorted_data = np.sort(data)
    cdf = np.arange(1, len(sorted_data) + 1) / len(sorted_data)
    ax3.plot(sorted_data, cdf, linewidth=2.6, color=HC['c1'], label='CDF')
    ax3.fill_between(sorted_data, cdf, alpha=0.20, color=HC['c1'])
    for p, label in zip([5, 25, 50, 75, 95], ['5th', '25th', '50th', '75th', '95th']):
        value = np.percentile(data, p)
        cdf_value = np.interp(value, sorted_data, cdf)
        ax3.plot(value, cdf_value, 'o', color=HC['c2'], markersize=8, markeredgecolor='white',
                  markeredgewidth=1.5, zorder=5)
        ax3.text(value, cdf_value + 0.04, f'{label}: {value:.2f}', fontweight='bold', fontsize=9, ha='center',
                  bbox=dict(boxstyle='round,pad=0.2', facecolor='white', edgecolor='#DDDDDD', alpha=0.9))
    ax3.set_xlabel(target_var, fontweight='bold'); ax3.set_ylabel('Cumulative Probability', fontweight='bold')
    ax3.set_title('(c) Cumulative Distribution', fontweight='bold', fontsize=14); despine(ax3)

    ax4 = axes[1, 1]; ax4.axis('off')
    stats_text = (
        f"TARGET VARIABLE STATISTICS\n{'-'*32}\n"
        f"Variable            {target_var}\nCount               {len(data):,}\n"
        f"Mean                {data.mean():.4f}\nStd Deviation       {data.std():.4f}\n"
        f"Variance            {data.var():.4f}\n{'-'*32}\n"
        f"Minimum             {data.min():.4f}\n5th Percentile      {np.percentile(data,5):.4f}\n"
        f"25th Percentile     {np.percentile(data,25):.4f}\nMedian              {data.median():.4f}\n"
        f"75th Percentile     {np.percentile(data,75):.4f}\n95th Percentile     {np.percentile(data,95):.4f}\n"
        f"Maximum             {data.max():.4f}\n{'-'*32}\n"
        f"IQR                 {np.percentile(data,75)-np.percentile(data,25):.4f}\n"
        f"Skewness            {stats.skew(data):.4f}\nKurtosis            {stats.kurtosis(data):.4f}\n"
    )
    ax4.text(0.05, 0.95, stats_text, transform=ax4.transAxes, verticalalignment='top', fontsize=12,
              fontweight='bold', bbox=dict(boxstyle='round', facecolor='#F6F8FA', edgecolor=HC['ink'], linewidth=1.6),
              family='monospace')

    plt.suptitle(f'Distribution of Target Variable: {target_var}  (n = {len(data):,})',
                 fontsize=17, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.savefig(output_dir / 'Target_Distribution.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Target Distribution")


def create_input_distributions():
    print("\nCreating Input Distributions...")
    plot_vars = [v for v in available_inputs if pd.api.types.is_numeric_dtype(df[v])]
    if not plot_vars:
        print("No numeric input variables"); return
    n_cols = 4
    n_rows = int(np.ceil(len(plot_vars) / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 4.5*n_rows), dpi=500)
    axes = np.array(axes).flatten()

    for idx, var in enumerate(plot_vars):
        ax = axes[idx]
        data = df[var].dropna()
        if len(data) == 0:
            continue
        color = PALETTE[idx % len(PALETTE)]
        ax.hist(data, bins=25, density=True, alpha=0.88, color=color, edgecolor=HC['ink'], linewidth=0.8)
        try:
            kde = stats.gaussian_kde(data)
            x_vals = np.linspace(data.min(), data.max(), 200)
            ax.plot(x_vals, kde(x_vals), color=HC['ink'], linewidth=2.2)
        except Exception:
            pass
        ax.axvline(data.mean(), color=HC['mean'], linestyle='-', linewidth=2.0, label=f'Mean: {data.mean():.2f}')
        ax.axvline(data.median(), color=HC['ink'], linestyle='--', linewidth=2.0, label=f'Median: {data.median():.2f}')
        clean_label = clean_column_name(var)
        ax.set_xlabel(clean_label, fontweight='bold', fontsize=11); ax.set_ylabel('Density', fontweight='bold', fontsize=11)
        ax.set_title(f'({chr(97+idx)}) {clean_label}', fontweight='bold', fontsize=12)
        ax.legend(loc='upper right', fontsize=8); despine(ax)
        add_n_label(ax, len(data), 'lower right')

    for idx in range(len(plot_vars), len(axes)):
        fig.delaxes(axes[idx])
    plt.suptitle('Distribution of All Input Variables', fontsize=17, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.savefig(output_dir / 'Input_Distributions.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Input Distributions")


def create_ridge_plot():
    print("\nCreating Ridge Plot of Input Distributions...")
    numeric_cols = [c for c in available_inputs if pd.api.types.is_numeric_dtype(df[c])]
    if len(numeric_cols) < 2:
        print("Not enough numeric variables for ridge plot"); return
    norm_df = pd.DataFrame({c: (df[c] - df[c].mean()) / df[c].std() for c in numeric_cols}).dropna()
    variables = list(norm_df.columns)[::-1]
    n = len(variables)

    fig, ax = plt.subplots(figsize=(10, max(6, n * 0.9)), dpi=500)
    x_grid = np.linspace(norm_df.values.min() - 1, norm_df.values.max() + 1, 300)
    overlap = 0.65
    for i, var in enumerate(variables):
        data = norm_df[var].dropna()
        color = PALETTE[i % len(PALETTE)]
        try:
            kde = stats.gaussian_kde(data)
            y = kde(x_grid)
            y = y / y.max()
        except Exception:
            continue
        baseline = i * overlap
        ax.fill_between(x_grid, baseline, baseline + y, color=color, alpha=0.85, linewidth=0, zorder=n - i)
        ax.plot(x_grid, baseline + y, color=HC['ink'], linewidth=1.4, zorder=n - i + 0.5)
        ax.text(x_grid.min() - 0.2, baseline + 0.05, clean_column_name(var), ha='right', va='bottom',
                 fontweight='bold', fontsize=10.5, color=HC['ink'])
    ax.set_yticks([])
    ax.set_xlabel('Standardized value (z-score)', fontweight='bold', fontsize=13)
    ax.set_title(f'Ridge Plot: Standardized Distributions of Input Variables  (n = {len(norm_df):,})',
                 fontweight='bold', fontsize=15, pad=15)
    for side in ['top', 'right', 'left']:
        ax.spines[side].set_visible(False)
    ax.spines['bottom'].set_color(HC['ink'])
    ax.grid(False)
    ax.set_xlim(x_grid.min() - 2.2, x_grid.max())
    plt.tight_layout()
    plt.savefig(output_dir / 'Ridge_Plot.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Ridge_Plot.png")


def _sens_slope(x, y, max_pairs=200000):
    n = len(x)
    idx_i, idx_j = np.triu_indices(n, k=1)
    n_pairs = len(idx_i)
    if n_pairs > max_pairs:
        rng = np.random.RandomState(42)
        keep = rng.choice(n_pairs, size=max_pairs, replace=False)
        idx_i, idx_j = idx_i[keep], idx_j[keep]
    dx = x[idx_j] - x[idx_i]
    dy = y[idx_j] - y[idx_i]
    valid = dx != 0
    slopes = dy[valid] / dx[valid]
    return np.median(slopes) if len(slopes) else 0.0


def _mann_kendall(y):
    n = len(y)
    s = 0.0
    for k in range(n - 1):
        s += np.sum(np.sign(y[k+1:] - y[k]))
    _, counts = np.unique(y, return_counts=True)
    tie_correction = np.sum(counts * (counts - 1) * (2 * counts + 5))
    var_s = (n * (n - 1) * (2 * n + 5) - tie_correction) / 18.0
    if var_s <= 0:
        return s, 0.0, 1.0
    if s > 0:
        z = (s - 1) / np.sqrt(var_s)
    elif s < 0:
        z = (s + 1) / np.sqrt(var_s)
    else:
        z = 0.0
    p = 2 * (1 - stats.norm.cdf(abs(z)))
    return s, z, p


def create_individual_parameter_plots():
    print("\nCreating Individual Parameter Distribution Plots...")
    all_vars = [target_var] + available_inputs
    numeric_vars = [v for v in all_vars if v in df.columns and pd.api.types.is_numeric_dtype(df[v])]
    if not numeric_vars:
        print("No numeric variables found"); return
    param_dir = output_dir / 'Individual_Parameter_Distributions'
    param_dir.mkdir(exist_ok=True, parents=True)

    for i, var in enumerate(numeric_vars):
        data = df[var].dropna()
        if len(data) < 5:
            continue
        color = PALETTE[i % len(PALETTE)]
        is_target = (var == target_var)

        fig = plt.figure(figsize=(9, 7), dpi=500)
        gs = fig.add_gridspec(2, 1, height_ratios=[1, 6], hspace=0.04)
        ax_box = fig.add_subplot(gs[0])
        ax_main = fig.add_subplot(gs[1], sharex=ax_box)

        bp = ax_box.boxplot(data, vert=False, patch_artist=True, widths=0.7)
        bp['boxes'][0].set_facecolor(color); bp['boxes'][0].set_alpha(0.85)
        bp['boxes'][0].set_edgecolor(HC['ink']); bp['boxes'][0].set_linewidth(1.4)
        bp['medians'][0].set_color('white'); bp['medians'][0].set_linewidth(2.2)
        for w in bp['whiskers'] + bp['caps']:
            w.set_color(HC['ink']); w.set_linewidth(1.3)
        bp['fliers'][0].set_marker('o'); bp['fliers'][0].set_markerfacecolor(HC['c2'])
        bp['fliers'][0].set_markeredgecolor('white'); bp['fliers'][0].set_markersize(6)
        ax_box.set_yticks([]); ax_box.axis('off')

        ax_main.hist(data, bins=30, density=True, alpha=0.65, color=color,
                      edgecolor=HC['ink'], linewidth=0.9)
        try:
            kde = stats.gaussian_kde(data)
            x_vals = np.linspace(data.min(), data.max(), 250)
            ax_main.plot(x_vals, kde(x_vals), color=HC['ink'], linewidth=2.6, label='KDE')
            ax_main.fill_between(x_vals, kde(x_vals), alpha=0.12, color=color)
        except Exception:
            pass

        mean_v, median_v, std_v = data.mean(), data.median(), data.std()
        ax_main.axvline(mean_v, color=HC['mean'], linestyle='-', linewidth=2.4, label=f'Mean = {mean_v:.3f}')
        ax_main.axvline(median_v, color=HC['c3'], linestyle='--', linewidth=2.4, label=f'Median = {median_v:.3f}')
        ax_main.axvspan(mean_v - std_v, mean_v + std_v, alpha=0.10, color=HC['neutral'], label='\u00b1 1 SD')

        clean_label = clean_column_name(var)
        ax_main.set_xlabel(clean_label + (' (Target)' if is_target else ''), fontweight='bold', fontsize=13)
        ax_main.set_ylabel('Probability Density', fontweight='bold', fontsize=13)
        ax_main.legend(loc='upper right', fontsize=10)
        despine(ax_main)

        stats_str = (f"n = {len(data):,}   |   SD = {std_v:.3f}   |   Skew = {stats.skew(data):.3f}   "
                     f"|   Kurtosis = {stats.kurtosis(data):.3f}")
        ax_main.text(0.02, 0.98, stats_str, transform=ax_main.transAxes, va='top', fontsize=9.5,
                      fontweight='bold', bbox=dict(boxstyle='round,pad=0.35', facecolor='white',
                                                    edgecolor=HC['ink'], linewidth=1.1, alpha=0.95))

        title_tag = 'Target Variable' if is_target else 'Input Parameter'
        fig.suptitle(f'{clean_label} \u2014 Distribution ({title_tag})', fontweight='bold', fontsize=15, y=0.965)
        plt.savefig(param_dir / f'{_safe_filename(var)}_Distribution.png', dpi=500,
                    bbox_inches='tight', facecolor='white')
        plt.close()
    print(f"Saved {len(numeric_vars)} individual parameter distribution plots to '{param_dir.name}/'")


def create_individual_trend_plots():
    print("\nCreating Individual Parameter Trend Plots...")
    all_vars = [target_var] + available_inputs
    numeric_vars = [v for v in all_vars if v in df.columns and pd.api.types.is_numeric_dtype(df[v])]
    if not numeric_vars:
        print("No numeric variables found"); return
    trend_dir = output_dir / 'Individual_Parameter_Trends'
    trend_dir.mkdir(exist_ok=True, parents=True)
    trend_summary = []

    for i, var in enumerate(numeric_vars):
        series = df[var].dropna()
        if len(series) < 8:
            continue
        x = np.arange(len(series))
        y = series.values
        color = PALETTE[i % len(PALETTE)]
        is_target = (var == target_var)

        window = max(5, len(y) // 15)
        roll_mean = pd.Series(y).rolling(window, center=True, min_periods=1).mean().values
        roll_std = pd.Series(y).rolling(window, center=True, min_periods=1).std().fillna(0).values

        sen_slope = _sens_slope(x.astype(float), y.astype(float))
        sen_intercept = np.median(y) - sen_slope * np.median(x)
        mk_s, mk_z, mk_p = _mann_kendall(y)
        if mk_p < 0.05:
            direction = 'Increasing' if sen_slope > 0 else 'Decreasing'
        else:
            direction = 'No significant trend'

        fig, ax = plt.subplots(figsize=(11, 6), dpi=500)
        sc = ax.scatter(x, y, c=y, cmap=SEQUENTIAL_CMAP, s=34, alpha=0.9,
                          edgecolors=HC['ink'], linewidth=0.5, zorder=3)
        ax.plot(x, roll_mean, color=HC['ink'], linewidth=2.4, label=f'Rolling mean (window={window})', zorder=4)
        ax.fill_between(x, roll_mean - roll_std, roll_mean + roll_std, color=HC['neutral'], alpha=0.15, zorder=1)

        trend_line = sen_intercept + sen_slope * x
        ax.plot(x, trend_line, color=HC['c2'], linewidth=2.4, linestyle='--',
                 label=f"Sen's slope trend ({direction})", zorder=4)

        clean_label = clean_column_name(var)
        ax.set_xlabel('Sample Index', fontweight='bold', fontsize=13)
        ax.set_ylabel(clean_label, fontweight='bold', fontsize=13)
        title_tag = 'Target Variable' if is_target else 'Input Parameter'
        ax.set_title(f'{clean_label} \u2014 Sequential Trend ({title_tag})', fontweight='bold', fontsize=15)
        ax.legend(loc='best', fontsize=9.5)
        despine(ax)

        stats_str = (f"n = {len(y):,}   |   Sen's slope = {sen_slope:.4g} / sample   |   Mann-Kendall Z = {mk_z:.2f}, "
                     f"p = {mk_p:.4f}")
        ax.text(0.02, 0.02, stats_str, transform=ax.transAxes, va='bottom', fontsize=9.5,
                 fontweight='bold', bbox=dict(boxstyle='round,pad=0.35', facecolor='white',
                                               edgecolor=HC['ink'], linewidth=1.1, alpha=0.95))
        cbar = fig.colorbar(sc, ax=ax, pad=0.01)
        cbar.set_label(clean_label, fontweight='bold', fontsize=10)

        plt.tight_layout()
        plt.savefig(trend_dir / f'{_safe_filename(var)}_Trend.png', dpi=500,
                    bbox_inches='tight', facecolor='white')
        plt.close()

        trend_summary.append({
            'Variable': clean_label, 'n': len(y), 'Sens_Slope': sen_slope,
            'Mann_Kendall_Z': mk_z, 'Mann_Kendall_p': mk_p, 'Trend_Direction': direction
        })

    if trend_summary:
        pd.DataFrame(trend_summary).to_csv(trend_dir / 'Trend_Statistics_Summary.csv', index=False)
    print(f"Saved {len(trend_summary)} individual parameter trend plots to '{trend_dir.name}/'")


def create_correlation_heatmap():
    print("\nCreating Correlation Heatmap...")
    numeric_vars = [v for v in ([target_var] + available_inputs)
                    if v in df.columns and pd.api.types.is_numeric_dtype(df[v])]
    corr_data = df[numeric_vars].dropna()
    if len(corr_data) < 5:
        print("Not enough data for correlation heatmap"); return
    corr_matrix = corr_data.corr()
    clean_labels = [clean_column_name(c) for c in corr_matrix.columns]
    n_vars = len(corr_matrix.columns)
    fig_size = max(12, n_vars * 0.7)

    fig, ax = plt.subplots(figsize=(fig_size, fig_size * 0.85), dpi=500)
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
    sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f', cmap=DIVERGING_CMAP, center=0,
                square=True, linewidths=1.4, linecolor='white',
                cbar_kws={"shrink": 0.8, "label": "Pearson Correlation"},
                annot_kws={'size': 10, 'weight': 'bold', 'color': HC['ink']}, ax=ax, vmin=-1, vmax=1)
    ax.set_xticklabels(clean_labels, rotation=45, ha='right', fontweight='bold', fontsize=11)
    ax.set_yticklabels(clean_labels, rotation=0, fontweight='bold', fontsize=11)
    ax.set_title(f'Pearson Correlation Matrix\nTarget: {target_var}   (n = {len(corr_data):,})',
                 fontsize=16, fontweight='bold', pad=20)
    plt.tight_layout()
    plt.savefig(output_dir / 'Correlation_Heatmap.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Correlation Heatmap")


def create_feature_target_plots():
    print("\nCreating Feature-Target Relationships...")
    plot_features = [v for v in available_inputs if pd.api.types.is_numeric_dtype(df[v])]
    if not plot_features:
        print("No numeric features"); return
    n_cols = 3
    n_rows = int(np.ceil(len(plot_features) / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 5*n_rows), dpi=500)
    axes = np.array(axes).flatten()

    for idx, feature in enumerate(plot_features):
        ax = axes[idx]
        data = df[[feature, target_var]].dropna()
        if len(data) < 10:
            continue
        x, y = data[feature], data[target_var]
        color = PALETTE[idx % len(PALETTE)]
        ax.scatter(x, y, alpha=0.8, s=34, c=color, edgecolors=HC['ink'], linewidth=0.5)

        z = np.polyfit(x, y, 1); p = np.poly1d(z)
        x_line = np.linspace(x.min(), x.max(), 100); y_pred = p(x_line)
        n = len(x); dof = n - 2; resid = y - p(x)
        se = np.sqrt(np.sum(resid**2) / dof); t_val = stats.t.ppf(0.975, dof)
        ci = t_val * se * np.sqrt(1/n + (x_line - x.mean())**2 / np.sum((x - x.mean())**2))

        ax.plot(x_line, y_pred, color=HC['c2'], linewidth=2.6, label='Linear Trend')
        ax.fill_between(x_line, y_pred - ci, y_pred + ci, alpha=0.25, color=HC['c2'], label='95% CI')

        pearson_corr, _ = pearsonr(x, y); spearman_corr, _ = spearmanr(x, y)
        ax.text(0.02, 0.98, f'Pearson: {pearson_corr:.3f}\nSpearman: {spearman_corr:.3f}\nn = {n:,}',
                 transform=ax.transAxes, va='top', fontweight='bold', fontsize=10,
                 bbox=dict(boxstyle='round', facecolor='white', alpha=0.95, edgecolor=HC['ink'], linewidth=1.1))

        clean_label = clean_column_name(feature)
        ax.set_xlabel(clean_label, fontweight='bold'); ax.set_ylabel(target_var, fontweight='bold')
        ax.set_title(f'({chr(97+idx)}) {clean_label}', fontweight='bold', fontsize=13)
        ax.legend(loc='best', fontsize=9); despine(ax)

    for idx in range(len(plot_features), len(axes)):
        fig.delaxes(axes[idx])
    plt.suptitle(f'Relationship Between Input Variables and {target_var}',
                 fontsize=17, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.savefig(output_dir / 'Feature_Target_Relationships.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Feature-Target Relationships")


def create_data_range_plot():
    print("\nCreating Data Range and Statistics...")
    all_vars = [target_var] + available_inputs
    numeric_vars = [v for v in all_vars if v in df.columns and pd.api.types.is_numeric_dtype(df[v])]
    stats_data = []
    for var in numeric_vars:
        data = df[var].dropna()
        if len(data) > 5:
            stats_data.append({'Variable': var, 'Min': data.min(), 'Q1': np.percentile(data, 25),
                                'Median': data.median(), 'Mean': data.mean(), 'Q3': np.percentile(data, 75),
                                'Max': data.max(), 'Std': data.std(), 'Count': len(data)})
    if not stats_data:
        print("Not enough data for range plot"); return
    stats_df = pd.DataFrame(stats_data).sort_values('Mean', ascending=False)
    clean_names = [clean_column_name(v) for v in stats_df['Variable']]

    fig, ax = plt.subplots(figsize=(14, max(8, len(stats_df)*0.35)), dpi=500)
    y_pos = np.arange(len(stats_df))
    ax.hlines(y_pos, stats_df['Min'], stats_df['Max'], color='#888888', linewidth=5, alpha=0.9)
    ax.hlines(y_pos, stats_df['Q1'], stats_df['Q3'], color=HC['c1'], linewidth=12, alpha=0.95, label='IQR')
    ax.scatter(stats_df['Mean'], y_pos, color=HC['c2'], s=130, zorder=5, edgecolors=HC['ink'],
                linewidth=1.6, label='Mean', marker='D')
    ax.scatter(stats_df['Median'], y_pos, color=HC['ink'], s=100, zorder=5, edgecolors='white',
                linewidth=1.4, label='Median', marker='s')
    ax.set_yticks(y_pos); ax.set_yticklabels(clean_names, fontweight='bold', fontsize=11)
    ax.set_xlabel('Value Range', fontweight='bold', fontsize=13)
    ax.set_title('Data Range and Statistics Summary\n(Min\u2013Max, IQR, Mean, Median)',
                 fontsize=16, fontweight='bold')
    ax.legend(loc='lower right'); despine(ax)
    for i, row in stats_df.iterrows():
        ax.text(1.02 * row['Max'], stats_df.index.get_loc(i), f'n={row["Count"]}', va='center',
                 fontweight='bold', fontsize=10)
    plt.tight_layout()
    plt.savefig(output_dir / 'Data_Range_Statistics.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Data Range and Statistics")


def create_qq_plots():
    print("\nCreating Q-Q Plots...")
    all_vars = [target_var] + available_inputs
    qq_vars = [v for v in all_vars if v in df.columns and pd.api.types.is_numeric_dtype(df[v])]
    if not qq_vars:
        print("No numeric variables for Q-Q plots"); return
    n_cols = 4
    n_rows = int(np.ceil(len(qq_vars) / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 4*n_rows), dpi=500)
    axes = np.array(axes).flatten()
    for idx, var in enumerate(qq_vars):
        ax = axes[idx]
        data = df[var].dropna()
        if len(data) < 10:
            continue
        (osm, osr), (slope, intercept, r) = stats.probplot(data, dist='norm', fit=True)
        color = PALETTE[idx % len(PALETTE)]
        ax.scatter(osm, osr, s=22, alpha=0.85, c=color, edgecolors=HC['ink'], linewidth=0.5)
        ax.plot(osm, slope * osm + intercept, color=HC['c2'], linewidth=2.4, label=f'R\u00b2 = {r**2:.3f}')
        clean_label = clean_column_name(var)
        ax.set_xlabel('Theoretical Quantiles', fontweight='bold', fontsize=11)
        ax.set_ylabel('Sample Quantiles', fontweight='bold', fontsize=11)
        ax.set_title(f'({chr(97+idx)}) {clean_label}  (n={len(data):,})', fontweight='bold', fontsize=12)
        ax.legend(loc='upper left', fontsize=9); despine(ax)
    for idx in range(len(qq_vars), len(axes)):
        fig.delaxes(axes[idx])
    plt.suptitle('Q-Q Plots for Normality Assessment', fontsize=17, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.savefig(output_dir / 'QQ_Plots.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Q-Q Plots")


def create_correlation_summary():
    print("\nCreating Correlation Summary...")
    numeric_vars = [v for v in ([target_var] + available_inputs)
                    if v in df.columns and pd.api.types.is_numeric_dtype(df[v])]
    corr_data = df[numeric_vars].dropna()
    if len(corr_data) < 5:
        print("Not enough data for correlation summary"); return
    corr_matrix = corr_data.corr()
    target_corr = corr_matrix[target_var].drop(target_var).sort_values(ascending=False)
    summary_df = pd.DataFrame({'Variable': target_corr.index, 'Correlation with Target': target_corr.values})
    summary_df['Variable'] = summary_df['Variable'].apply(clean_column_name)
    summary_df.to_csv(output_dir / 'Correlation_Summary.csv', index=False)
    with open(output_dir / 'Correlation_Summary.txt', 'w') as f:
        f.write("=" * 70 + f"\nCORRELATION WITH TARGET VARIABLE\nTarget: {target_var}\n" + "=" * 70 + "\n\n")
        for _, row in summary_df.iterrows():
            f.write(f"{row['Variable']:<40} {row['Correlation with Target']:>8.4f}\n")
    fig, ax = plt.subplots(figsize=(12, max(6, len(summary_df)*0.45)), dpi=500)
    colors = [HC['c2'] if v < 0 else HC['c1'] for v in summary_df['Correlation with Target']]
    bars = ax.barh(summary_df['Variable'], summary_df['Correlation with Target'], color=colors,
                     edgecolor=HC['ink'], linewidth=1.3)
    ax.axvline(0, color=HC['ink'], linewidth=1.4)
    ax.set_xlabel(f'Pearson Correlation with {target_var}', fontweight='bold')
    ax.set_title(f'Feature Correlation with {target_var}  (n = {len(corr_data):,})',
                 fontweight='bold', fontsize=14)
    ax.invert_yaxis(); despine(ax)
    for bar, val in zip(bars, summary_df['Correlation with Target']):
        offset = 0.02 if val >= 0 else -0.02
        ha = 'left' if val >= 0 else 'right'
        ax.text(val + offset, bar.get_y() + bar.get_height()/2, f'{val:.3f}', va='center', ha=ha,
                 fontweight='bold', fontsize=10)
    plt.tight_layout()
    plt.savefig(output_dir / 'Correlation_with_Target_Bar.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Correlation Summary")


def create_missing_value_map():
    print("\nCreating Missing Value Map...")
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    if len(missing) == 0:
        print("No missing values found")
        with open(output_dir / 'Missing_Values_Report.txt', 'w') as f:
            f.write("MISSING VALUES REPORT\n" + "=" * 50 + "\n")
            f.write("No missing values detected in the dataset.\n")
            f.write(f"Total rows: {len(df)}\nTotal columns: {len(df.columns)}\n")
        return
    missing_pct = (missing / len(df)) * 100
    missing_df = pd.DataFrame({'Missing': missing, 'Percent': missing_pct}).sort_values('Missing', ascending=False)
    missing_df.to_csv(output_dir / 'Missing_Values_Report.csv')
    fig, ax = plt.subplots(figsize=(12, max(6, len(missing_df)*0.5)), dpi=500)
    y_pos = np.arange(len(missing_df))
    bars = ax.barh(y_pos, missing_df['Missing'], color=HC['c2'], edgecolor=HC['ink'], linewidth=1.3)
    ax.set_yticks(y_pos); ax.set_yticklabels([clean_column_name(c) for c in missing_df.index],
                                              fontweight='bold', fontsize=11)
    ax.set_xlabel('Number of Missing Values', fontweight='bold')
    ax.set_title(f'Missing Values by Column  (Total n = {len(df):,})', fontweight='bold', fontsize=14)
    ax.invert_yaxis(); despine(ax)
    for bar, val, pct in zip(bars, missing_df['Missing'], missing_df['Percent']):
        ax.text(val + max(missing_df['Missing'])*0.01, bar.get_y() + bar.get_height()/2, f'{val} ({pct:.1f}%)',
                 va='center', fontweight='bold', fontsize=10)
    plt.tight_layout()
    plt.savefig(output_dir / 'Missing_Values_Map.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Missing Values Map")


def create_violin_plots():
    print("\nCreating Violin Plots...")
    numeric_cols = [c for c in available_inputs if pd.api.types.is_numeric_dtype(df[c])]
    if not numeric_cols:
        return
    n_cols = 4
    n_rows = int(np.ceil(len(numeric_cols) / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 4*n_rows), dpi=500)
    axes = np.array(axes).flatten()
    for idx, var in enumerate(numeric_cols):
        ax = axes[idx]
        data = df[var].dropna()
        if len(data) < 5:
            continue
        color = PALETTE[idx % len(PALETTE)]
        parts = ax.violinplot(data, showmeans=True, showmedians=True, widths=0.6)
        for pc in parts['bodies']:
            pc.set_facecolor(color); pc.set_alpha(0.85); pc.set_edgecolor(HC['ink']); pc.set_linewidth(1.3)
        for key in ['cmeans', 'cmedians', 'cbars', 'cmins', 'cmaxes']:
            if key in parts:
                parts[key].set_color(HC['c2'] if key == 'cmeans' else HC['ink'])
                parts[key].set_linewidth(2.0)
        clean_label = clean_column_name(var)
        ax.set_ylabel(clean_label, fontweight='bold', fontsize=11)
        ax.set_title(f'({chr(97+idx)}) {clean_label}  (n={len(data):,})', fontweight='bold', fontsize=12)
        ax.set_xticks([]); despine(ax)
    for idx in range(len(numeric_cols), len(axes)):
        fig.delaxes(axes[idx])
    plt.suptitle('Violin Plots of Input Variables', fontsize=17, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.savefig(output_dir / 'Violin_Plots.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Violin Plots")


def create_pair_plot():
    print("\nCreating Pair Plot...")
    numeric_cols = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]
    if len(numeric_cols) < 2:
        return
    pair_vars = [target_var] + [c for c in numeric_cols if c != target_var][:5]
    pair_vars = [v for v in pair_vars if v in df.columns]
    if len(pair_vars) < 2:
        return
    pair_df = df[pair_vars].dropna()
    if len(pair_df) < 5:
        return
    g = sns.pairplot(pair_df, diag_kind='kde', corner=True,
                      plot_kws={'alpha': 0.75, 's': 24, 'edgecolor': HC['ink'], 'linewidth': 0.5, 'color': HC['c1']},
                      diag_kws={'color': HC['c2'], 'linewidth': 2.4, 'fill': True, 'alpha': 0.4})
    g.fig.suptitle(f'Pair Plot of Target and Key Features  (n = {len(pair_df):,})',
                   fontweight='bold', fontsize=16, y=1.02)
    g.fig.savefig(output_dir / 'Pair_Plot.png', dpi=400, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Pair Plot")


def create_3d_scatter():
    print("\nCreating 3D Scatter Plot...")
    numeric_cols = [c for c in available_inputs if pd.api.types.is_numeric_dtype(df[c])]
    if len(numeric_cols) < 3:
        print("Not enough numeric variables for 3D scatter"); return
    fig = plt.figure(figsize=(12, 9), dpi=500)
    ax = fig.add_subplot(111, projection='3d')
    x_var, y_var, z_var = numeric_cols[0], numeric_cols[1], numeric_cols[2]
    data = df[[x_var, y_var, z_var, target_var]].dropna()
    if len(data) < 10:
        return
    sc = ax.scatter(data[x_var], data[y_var], data[z_var], c=data[target_var], cmap=SEQUENTIAL_CMAP,
                      s=40, alpha=0.9, edgecolors=HC['ink'], linewidth=0.5)
    ax.set_xlabel(clean_column_name(x_var), fontweight='bold', fontsize=11)
    ax.set_ylabel(clean_column_name(y_var), fontweight='bold', fontsize=11)
    ax.set_zlabel(clean_column_name(z_var), fontweight='bold', fontsize=11)
    ax.set_title(f'3D Scatter: Features Colored by {target_var}  (n = {len(data):,})',
                 fontweight='bold', fontsize=14)
    cbar = fig.colorbar(sc, ax=ax, shrink=0.6, pad=0.1)
    cbar.set_label(target_var, fontweight='bold', fontsize=12)
    plt.tight_layout()
    plt.savefig(output_dir / '3D_Scatter.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved 3D Scatter Plot")


def create_feature_importance():
    print("\nCreating Feature Importance (Correlation + Bootstrapped RF)...")
    try:
        from sklearn.ensemble import RandomForestRegressor
        from sklearn.preprocessing import StandardScaler
        SKLEARN_AVAILABLE = True
    except ImportError:
        SKLEARN_AVAILABLE = False
        print("scikit-learn not available - using correlation-based only")

    numeric_cols = [c for c in available_inputs if pd.api.types.is_numeric_dtype(df[c])]
    if not numeric_cols or target_var not in df.columns:
        return
    clean_df = df[numeric_cols + [target_var]].dropna()
    if len(clean_df) < 10:
        return
    X = clean_df[numeric_cols]; y = clean_df[target_var]
    corr_imp = pd.Series(np.abs(np.array([pearsonr(X[c], y)[0] for c in numeric_cols])),
                          index=numeric_cols).sort_values(ascending=True)

    rf_imp_mean, rf_imp_std = None, None
    if SKLEARN_AVAILABLE:
        try:
            scaler = StandardScaler()
            X_scaled = scaler.fit_transform(X)
            n_boot = 50
            boot_importances = np.zeros((n_boot, len(numeric_cols)))
            rng = np.random.RandomState(42)
            for b in range(n_boot):
                idx = rng.choice(len(X_scaled), size=len(X_scaled), replace=True)
                rf = RandomForestRegressor(n_estimators=150, random_state=b, n_jobs=-1)
                rf.fit(X_scaled[idx], y.values[idx])
                boot_importances[b, :] = rf.feature_importances_
            rf_imp_mean = pd.Series(boot_importances.mean(axis=0), index=numeric_cols).sort_values(ascending=True)
            rf_imp_std = pd.Series(boot_importances.std(axis=0), index=numeric_cols)[rf_imp_mean.index]
        except Exception as e:
            print(f"Random Forest bootstrap failed: {e}")

    n_plots = 2 if rf_imp_mean is not None else 1
    fig, axes = plt.subplots(1, n_plots, figsize=(10*n_plots, max(6, len(numeric_cols)*0.5)), dpi=500)
    if n_plots == 1:
        axes = [axes]
    ax1 = axes[0]
    bars1 = ax1.barh([clean_column_name(c) for c in corr_imp.index], corr_imp.values,
                       color=HC['c1'], edgecolor=HC['ink'], linewidth=1.3)
    ax1.set_xlabel('|Pearson Correlation|', fontweight='bold')
    ax1.set_title('(a) Correlation-based Importance', fontweight='bold', fontsize=14)
    despine(ax1)
    for bar, val in zip(bars1, corr_imp.values):
        ax1.text(val + max(corr_imp.values)*0.01, bar.get_y() + bar.get_height()/2, f'{val:.3f}',
                   va='center', fontweight='bold', fontsize=10)
    if rf_imp_mean is not None:
        ax2 = axes[1]
        ax2.barh([clean_column_name(c) for c in rf_imp_mean.index], rf_imp_mean.values,
                   xerr=rf_imp_std.values, color=HC['c2'], edgecolor=HC['ink'], linewidth=1.3,
                   error_kw=dict(ecolor=HC['ink'], capsize=4, elinewidth=1.8, capthick=1.8))
        ax2.set_xlabel('Random Forest Importance (mean \u00b1 SD, 50 bootstraps)', fontweight='bold')
        ax2.set_title('(b) Bootstrapped RF Feature Importance', fontweight='bold', fontsize=14)
        despine(ax2)
    plt.suptitle(f'Feature Importance for Predicting {target_var}  (n = {len(clean_df):,})',
                 fontsize=17, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.savefig(output_dir / 'Feature_Importance.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()

    imp_df = pd.DataFrame({'Variable': corr_imp.index, 'Correlation_Importance': corr_imp.values})
    if rf_imp_mean is not None:
        imp_df['RF_Importance_Mean'] = rf_imp_mean.values
        imp_df['RF_Importance_SD'] = rf_imp_std.values
    imp_df.to_csv(output_dir / 'Feature_Importance.csv', index=False)
    print("Saved Feature Importance")


def create_effect_size_forest_plot():
    print("\nCreating Effect-Size Forest Plot...")
    try:
        import statsmodels.api as sm
    except ImportError:
        print("statsmodels not available - skipping forest plot"); return
    numeric_cols = [c for c in available_inputs if pd.api.types.is_numeric_dtype(df[c])]
    if len(numeric_cols) < 2:
        print("Not enough numeric variables for forest plot"); return
    clean_df = df[numeric_cols + [target_var]].dropna()
    if len(clean_df) < 10:
        return
    Xz = (clean_df[numeric_cols] - clean_df[numeric_cols].mean()) / clean_df[numeric_cols].std()
    yz = (clean_df[target_var] - clean_df[target_var].mean()) / clean_df[target_var].std()
    Xz_const = sm.add_constant(Xz)
    model = sm.OLS(yz, Xz_const).fit()
    coefs = model.params.drop('const')
    conf = model.conf_int().drop('const')
    pvals = model.pvalues.drop('const')
    order = coefs.sort_values().index
    coefs, conf, pvals = coefs[order], conf.loc[order], pvals[order]

    fig, ax = plt.subplots(figsize=(10, max(6, len(coefs)*0.55)), dpi=500)
    y_pos = np.arange(len(coefs))
    colors = [HC['c2'] if p < 0.05 else HC['neutral'] for p in pvals]
    ax.hlines(y_pos, conf[0], conf[1], color=colors, linewidth=2.8, alpha=0.95)
    ax.scatter(coefs.values, y_pos, color=colors, s=95, zorder=5, edgecolors=HC['ink'], linewidth=1.5)
    ax.axvline(0, color=HC['ink'], linewidth=1.4, linestyle='--')
    ax.set_yticks(y_pos)
    ax.set_yticklabels([clean_column_name(c) for c in coefs.index], fontweight='bold', fontsize=11)
    ax.set_xlabel('Standardized Coefficient (95% CI)', fontweight='bold', fontsize=13)
    ax.set_title(f'Effect Sizes on {target_var}\n(Standardized OLS Coefficients)',
                 fontweight='bold', fontsize=15)
    despine(ax)
    legend_elements = [
        Line2D([0], [0], marker='o', color='w', markerfacecolor=HC['c2'], markersize=10, label='p < 0.05'),
        Line2D([0], [0], marker='o', color='w', markerfacecolor=HC['neutral'], markersize=10, label='p \u2265 0.05'),
    ]
    ax.legend(handles=legend_elements, loc='best')
    fig.text(0.02, 0.01, f"Model R\u00b2 = {model.rsquared:.3f}   |   n = {len(clean_df):,}",
             fontsize=11, fontweight='bold')
    plt.tight_layout(rect=[0, 0.035, 1, 1])
    plt.savefig(output_dir / 'Effect_Size_Forest_Plot.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    coef_df = pd.DataFrame({'Variable': coefs.index, 'Std_Coefficient': coefs.values,
                             'CI_lower': conf[0].values, 'CI_upper': conf[1].values, 'p_value': pvals.values})
    coef_df.to_csv(output_dir / 'Effect_Size_Coefficients.csv', index=False)
    print("Saved Effect_Size_Forest_Plot.png")


def create_residual_diagnostics():
    print("\nCreating Residual Diagnostics...")
    try:
        import statsmodels.api as sm
    except ImportError:
        print("statsmodels not available - skipping residual diagnostics"); return
    numeric_cols = [c for c in available_inputs if pd.api.types.is_numeric_dtype(df[c])]
    if not numeric_cols:
        return
    clean_df = df[numeric_cols + [target_var]].dropna()
    if len(clean_df) < 10:
        return
    X = sm.add_constant(clean_df[numeric_cols])
    y = clean_df[target_var]
    model = sm.OLS(y, X).fit()
    fitted = model.fittedvalues
    residuals = model.resid
    std_resid = (residuals - residuals.mean()) / residuals.std()

    fig, axes = plt.subplots(2, 2, figsize=(13, 10), dpi=500)
    ax1 = axes[0, 0]
    ax1.scatter(fitted, residuals, alpha=0.8, s=34, c=HC['c1'], edgecolors=HC['ink'], linewidth=0.5)
    ax1.axhline(0, color=HC['c2'], linewidth=2.2, linestyle='--')
    ax1.set_xlabel('Fitted Values', fontweight='bold'); ax1.set_ylabel('Residuals', fontweight='bold')
    ax1.set_title('(a) Residuals vs Fitted', fontweight='bold', fontsize=13); despine(ax1)

    ax2 = axes[0, 1]
    (osm, osr), (slope, intercept, r) = stats.probplot(std_resid, dist='norm', fit=True)
    ax2.scatter(osm, osr, s=26, alpha=0.85, c=HC['c1'], edgecolors=HC['ink'], linewidth=0.5)
    ax2.plot(osm, slope*osm + intercept, color=HC['c2'], linewidth=2.4, label=f'R\u00b2 = {r**2:.3f}')
    ax2.set_xlabel('Theoretical Quantiles', fontweight='bold'); ax2.set_ylabel('Standardized Residuals', fontweight='bold')
    ax2.set_title('(b) Normal Q-Q of Residuals', fontweight='bold', fontsize=13)
    ax2.legend(loc='upper left'); despine(ax2)

    ax3 = axes[1, 0]
    ax3.hist(residuals, bins=30, density=True, alpha=0.85, color=HC['c3'], edgecolor=HC['ink'], linewidth=0.8)
    try:
        kde = stats.gaussian_kde(residuals)
        xv = np.linspace(residuals.min(), residuals.max(), 200)
        ax3.plot(xv, kde(xv), color=HC['ink'], linewidth=2.4)
    except Exception:
        pass
    ax3.set_xlabel('Residual', fontweight='bold'); ax3.set_ylabel('Density', fontweight='bold')
    ax3.set_title('(c) Residual Distribution', fontweight='bold', fontsize=13); despine(ax3)

    ax4 = axes[1, 1]
    ax4.scatter(fitted, np.sqrt(np.abs(std_resid)), alpha=0.8, s=34, c=HC['c4'], edgecolors=HC['ink'], linewidth=0.5)
    ax4.set_xlabel('Fitted Values', fontweight='bold'); ax4.set_ylabel('\u221a|Standardized Residuals|', fontweight='bold')
    ax4.set_title('(d) Scale-Location Plot', fontweight='bold', fontsize=13); despine(ax4)

    plt.suptitle(f'Regression Diagnostics (OLS baseline, R\u00b2 = {model.rsquared:.3f}, n = {len(clean_df):,})',
                 fontsize=17, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.savefig(output_dir / 'Residual_Diagnostics.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Residual_Diagnostics.png")


def create_clustermap():
    print("\nCreating Correlation Clustermap...")
    numeric_vars = [v for v in ([target_var] + available_inputs)
                    if v in df.columns and pd.api.types.is_numeric_dtype(df[v])]
    if len(numeric_vars) < 3:
        return
    corr_data = df[numeric_vars].dropna()
    if len(corr_data) < 5:
        return
    corr_matrix = corr_data.corr()
    labels = [clean_column_name(c) for c in corr_matrix.columns]
    try:
        g = sns.clustermap(corr_matrix, annot=True, fmt='.2f', cmap=DIVERGING_CMAP, center=0,
                             linewidths=1.4, linecolor='white', annot_kws={'size': 9, 'weight': 'bold'},
                             figsize=(14, 12), dendrogram_ratio=0.15,
                             cbar_kws={'label': 'Pearson Correlation'})
        g.ax_heatmap.set_xticklabels(labels, rotation=45, ha='right', fontweight='bold', fontsize=10)
        g.ax_heatmap.set_yticklabels(labels, rotation=0, fontweight='bold', fontsize=10)
        g.fig.suptitle(f'Clustered Correlation Heatmap  (n = {len(corr_data):,})',
                       fontweight='bold', fontsize=16, y=1.02)
        g.fig.savefig(output_dir / 'Correlation_Clustermap.png', dpi=500, bbox_inches='tight', facecolor='white')
        plt.close()
        print("Saved Correlation Clustermap")
    except Exception as e:
        print(f"Clustermap failed: {e}")


def create_train_test_distribution_comparison():
    print("\nCreating Train-Test Distribution Comparison...")
    from sklearn.model_selection import train_test_split
    df_clean = df.dropna().reset_index(drop=True)
    train_df, test_df = train_test_split(df_clean, test_size=0.30, random_state=42, shuffle=True)
    numeric_cols = [c for c in df_clean.columns if pd.api.types.is_numeric_dtype(df_clean[c])]
    if not numeric_cols:
        return
    n_cols = 4
    n_rows = int(np.ceil(len(numeric_cols) / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 4*n_rows), dpi=500)
    axes = np.array(axes).flatten()
    for idx, var in enumerate(numeric_cols):
        ax = axes[idx]
        train_data = train_df[var].dropna(); test_data = test_df[var].dropna()
        if len(train_data) < 3 or len(test_data) < 3:
            continue
        ax.hist(train_data, bins=20, density=True, alpha=0.7, color=HC['c1'], edgecolor=HC['ink'],
                  linewidth=0.9, label=f'Train (n={len(train_data):,})')
        ax.hist(test_data, bins=20, density=True, alpha=0.7, color=HC['c2'], edgecolor=HC['ink'],
                  linewidth=0.9, label=f'Test (n={len(test_data):,})')
        ks_stat, ks_p = stats.ks_2samp(train_data, test_data)
        ax.set_xlabel(clean_column_name(var), fontweight='bold', fontsize=11)
        ax.set_ylabel('Density', fontweight='bold', fontsize=11)
        ax.set_title(f'({chr(97+idx)}) {clean_column_name(var)}\nKS p={ks_p:.3f}',
                     fontweight='bold', fontsize=12)
        ax.legend(fontsize=8); despine(ax)
    for idx in range(len(numeric_cols), len(axes)):
        fig.delaxes(axes[idx])
    plt.suptitle(f'Train vs Test Distribution Comparison (70:30, n = {len(df_clean):,})',
                 fontsize=17, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.savefig(output_dir / 'Train_Test_Distribution_Comparison.png', dpi=500, bbox_inches='tight', facecolor='white')
    plt.close()
    print("Saved Train-Test Distribution Comparison")


# =======================
# MAIN
# =======================
def create_all_figures():
    print("\n" + "=" * 70)
    print("3DP-FRC EDA PIPELINE  |  Target: " + target_var)
    print("=" * 70)
    print(f"Output directory: {output_dir}")
    print(f"Variables: {len(available_inputs)} inputs + 1 target")
    print(f"Total observations (n): {len(df):,}")

    # Primary deliverables requested by the user:
    df_clean = save_cleaned_dataset()
    table2 = create_table2_statistics(df_clean)

    # Extended descriptive stats (kept from original)
    create_descriptive_statistics()

    # Full figure suite (unchanged behaviour)
    create_dataset_composition()
    create_target_distribution()
    create_input_distributions()
    create_ridge_plot()
    create_individual_parameter_plots()
    create_individual_trend_plots()
    create_correlation_heatmap()
    create_feature_target_plots()
    create_data_range_plot()
    create_qq_plots()
    create_correlation_summary()
    create_missing_value_map()
    create_violin_plots()
    create_pair_plot()
    create_3d_scatter()
    create_pca_biplot()
    create_clustermap()
    create_feature_importance()
    create_effect_size_forest_plot()
    create_residual_diagnostics()
    create_train_test_split()
    create_train_test_distribution_comparison()

    print("\n" + "=" * 70)
    print("ALL DONE.")
    print(f"Location: {output_dir}")
    print("=" * 70)
    print("\nPrimary deliverables (requested):")
    print("   * Cleaned_Dataset.csv          (rows with any NaN dropped)")
    print("   * Cleaning_Log.txt             (per-column missing-rate audit)")
    print("   * Table2_Descriptive_Statistics.csv")
    print("   * Table2_Descriptive_Statistics.png  (publication-ready)")
    print("   * Table2_Descriptive_Statistics.txt")
    print("\nTable 2 columns: Variable | n | Missing (n) | Missing (%) |")
    print("                  Min | Q1 | Median | Mean | Std | Max")
    print("                  (Target 'CS (MPa)' is placed first and highlighted)")


if __name__ == "__main__":
    create_all_figures()
```
